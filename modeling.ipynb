# Моделирование на T-ECD
Сохранена исходная двухуровневая схема: implicit ALS формирует кандидатов, Random Forest использует ALS score и доступный контекст событий.

```python
from tecd_pipeline import load_events, temporal_split, train

DATA_DIR = "data/t_ecd"
events = load_events(DATA_DIR, max_rows=2_000_000)
train_events, test_events = temporal_split(events, test_days=2)
print("train days:", train_events.day.min(), train_events.day.max())
print("test days:", test_events.day.min(), test_events.day.max())
```

```python
metadata = train(DATA_DIR, max_rows=2_000_000)
metadata
```

## Интерпретация
`metadata` содержит результаты только текущей локальной выборки. Случайное разбиение не используется, чтобы будущие события не попадали в обучение. Прежние метрики банковского датасета неприменимы. Для ALS основными offline-метриками обычно являются Recall@K/NDCG@K, однако их добавление изменило бы функциональность проекта; поэтому здесь сохраняется имевшаяся оценка второго уровня и не заявляются результаты, не полученные запуском.
