# Загрузка T-ECD
Загружаем только нужные домены и последовательные дни T-ECD Small. Полный набор (2.81 ТБ) локально не требуется.

```python
import importlib.util
import os
import sys
from huggingface_hub import hf_hub_download

# Импортируем официальный downloader без копирования его кода в проект.
path = hf_hub_download("t-tech/T-ECD", "tecd_downloader.py", repo_type="dataset")
spec = importlib.util.spec_from_file_location("tecd_downloader", path)
tecd_downloader = importlib.util.module_from_spec(spec)
sys.modules[spec.name] = tecd_downloader
spec.loader.exec_module(tecd_downloader)
```

```python
tecd_downloader.download_dataset(
    token=os.environ["HF_TOKEN"],
    dataset_path="dataset/small",
    local_dir="data/t_ecd",
    domains=["retail", "marketplace", "offers"],
    day_begin=1250,
    day_end=1308,
    max_workers=10,
)
```
